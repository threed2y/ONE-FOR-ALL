Newton Raphson Method

```python
import math
import sys
import numpy as np
import matplotlib.pyplot as plt
```

```python
def newton_raphson(f, df, x0, tol=0.000001, max_iter=100):
    
    for i in range(max_iter):
        
        x1 = x0 - f(x0) / df(x0)
        
        print(i + 1, x1)
        
        if abs(x1 - x0) < tol:
            return x1
        
        x0 = x1
    
    return x0

def f(x):
    return x**2 - 4

def df(x):
    return 2*x

root = newton_raphson(f,df,1)
```

```python
esp = 0.000001

def fx(x):
    return math.exp(x) - 3 * x

def regula_falsi(N):
    x0 = float(input("Enter Initial Guess X0: "))
    x1 = float(input("Enter Initial Guess X1: "))
    
    f0 = fx(x0)
    f1 = fx(x1)

    if (f0 * f1) >= 0:
        print("Modify x0 or x1 as both signs are equal")
        sys.exit()
    if f0 > 0:
        x0, x1 = x1, x0
        f0, f1 = f1, f0
    for i in range(2, N):
        x2 = (x0 * f1 - f0 * x1)/(f1 - f0)
        print(f"x{i} at iteration {i-1} is {x2}")
        f2 = fx(x2)
        if abs(f2) < esp:
            print(f"{x2} is the approximate root of equation. \n")
            break
        elif f2 < 0:
            x0, f0 = x2, f2
        else:
            x1, f1 = x2, f2
    print(f"The final solution is: {x2}")

regula_falsi(50)
```

```python
def regula_falsi_2(f, a, b, tol=0.000001, max_iter=100):
    
    if f(a) * f(b) >= 0:
        print("Invalid interval")
        return
    
    for i in range(max_iter):
        
        c = (a * f(b) - b * f(a)) / (f(b) - f(a))
        
        print(i + 1, c)
        
        if abs(f(c)) < tol:
            return c
        
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    
    return c

def f(x):
    return math.exp(x) - 3 * x

root = regula_falsi_2(f, 1, 3)

print("Root =", root)
```
