<h2> Conversion from decimal to other number systems </h2>

```python
# Decimal to Octal    

n = int(input("enter a number: "))
octal = ""

if n == 0:
    print("0")

while n > 0:
    remainder = n % 8
    octal = str(remainder) + octal
    n = n // 8

print(octal)
```

```python
# Decimal to Binary
n = int(input())
binary_l = ""

if n == 0:
    print("0")

else:
    while n > 0:
        remainder = n % 2
        binary_l = binary_l + str(remainder) 
        n = n // 2

print(binary_l)
```

```python
# Decimal to hexadecimal Conversion
n = int(input())
hexa_decimal = {
    "A":10,
    "B":11,
    "C":12,
    "D":13,
    "E":14,
    "F":15
}

h_d = ""

if n == 0:
    print("0")

else:
    while n > 0:
        remainder = n % 16
        if remainder in hexa_decimal.keys():
            h_d = hexa_decimal[remainder] + h_d
        else:
            h_d = str(remainder) + h_d
            
        n = n // 16

print(h_d)
```

```python
# Fractional Decimal to Binary

# Real Part
number = str(input("enter a number: "))
binary_l = ""

real_part = int(number.split(".")[0])
frac_part = int(number.split(".")[1])

print(f"The real part of your number is {real_part}",
      f"The fractional part of your number is {frac_part}", sep="\n")


if int(real_part) == 0:
    print("0")
    
else:
    while int(real_part) > 0:
        remainder = int(real_part) % 2
        binary_l = binary_l + str(remainder) 
        real_part = real_part // 2

binary_l = binary_l[::-1]

# Fractional Part
fractional = ""

count = 10

while frac_part != 0 and count > 0:
    frac_part *= 2

    if frac_part >= 1:
        fractional += "1"
        frac_part -= 1
    else:
        fractional += "0"

    count -= 1

print(binary_l + "." + fractional)
```

```python
# fractional decimal to hexadecimal

number = float(input("Enter fractional decimal: "))

reference = "0123456789ABCDEF"

result = ""

fraction = number

for i in range(10):
    fraction *= 16

    value = int(fraction)

    result += reference[value]

    fraction -= value

    if fraction == 0:
        break

print("Hexadecimal:", "0." + result)
```

```python
# fractional decimal to octal

number = float(input("Enter fractional decimal: "))

fraction = number
result = ""

for i in range(10):
    fraction *= 8

    value = int(fraction)

    result += str(value)

    fraction -= value

    if fraction == 0:
        break

print("Octal:", "0." + result)
```
