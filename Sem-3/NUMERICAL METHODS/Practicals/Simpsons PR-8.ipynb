Simpson's 1/3 Rule

```python
def simpson_1_3(f, a, b, n):
    if n % 2 != 0:
        print("n must be even")
        return

    h = (b - a) / n

    y = []

    for i in range(n + 1):
        x = a + i * h
        y.append(f(x))

    so = 0
    se = 0

    for i in range(1, n):
        if i % 2 == 0:
            se = se + y[i]
        else:
            so = so + y[i]

    ans = (h / 3) * (y[0] + y[n] + 4 * so + 2 * se)

    return ans

f = lambda x: x**2

simpson_1_3(f,0,2,4)
```

Simpson's 3/8 Rule

```python
def simpson_3_8(f, a, b, n):
    if n % 3 != 0:
        print("n must be a multiple of 3")
        return

    h = (b - a) / n

    y = []

    for i in range(n + 1):
        x = a + i * h
        y.append(f(x))

    s3 = 0
    s2 = 0

    for i in range(1, n):
        if i % 3 == 0:
            s2 = s2 + y[i]
        else:
            s3 = s3 + y[i]

    ans = (3 * h / 8) * (y[0] + y[n] + 3 * s3 + 2 * s2)

    return ans

simpson_3_8(f,0,2,6)
```
