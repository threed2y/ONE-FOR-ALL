<h2>Subtraction of Binary numbers</h2>

```python
# Subtraction of two binary numbers

binary1, binary2 = input("Enter binary numbers separated by comma: ").split(",")

binary1 = binary1[::-1]
binary2 = binary2[::-1]

borrow = 0
answer = ""

for i, j in zip(binary1, binary2):

    total = int(i) - int(j) - borrow

    if total == -1:
        answer += "1"
        borrow = 1
    else:
        answer += str(total)
        borrow = 0

if borrow:
    print("Negative result")
else:
    answer = answer[::-1]
    print("Answer:", answer)
```

```python
# Subtraction of two fractional binary numbers

a, b = input("Enter two binary fractions: ").split(",")

a = a.replace(".", "")
b = b.replace(".", "")

borrow = 0
answer = ""

for i, j in zip(a[::-1], b[::-1]):
    difference = int(i) - int(j) - borrow

    if difference < 0:
        difference += 2
        borrow = 1
    else:
        borrow = 0

    answer += str(difference)

answer = answer[::-1]

print(answer)
```
