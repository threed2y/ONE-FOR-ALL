```python
import math

def trap(f, a, b, n):
    h = (b-a)/n
    s = 0

    for i in range(n):
        s += (f(a+i*h) + f(a+(i+1)*h))*h/2

    return s

def romberg(f, a, b, k):
    R = [[None]*k for _ in range(k)]

    for i in range(k):
        R[i][0] = trap(f, a, b, 2**i)

    for j in range(1, k):
        for i in range(k-j):
            R[i][j] = (4**j * R[i+1][j-1] - R[i][j-1])/(4**j - 1)
    return R

def f(x):
    return math.log(x)

answer = romberg(f, 1, 6, 4)

for row in answer:
    formatted_row = [f"{val:.5f}" for val in row if val is not None]
    print(formatted_row)

print(f"\nFinal Answer: {answer[0][-1]:.10f}")
```
