```python
def trapezoidal_rule(f, a, b, n):
    h = (b - a) / n

    total = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h
        total += 2 * f(x)

    integral = (h / 2) * total

    return integral

f = lambda x: x**2

result = trapezoidal_rule(f, 0, 2, 1000000)

print("Integral =", result)
```
