```python
# Gram Schmidt Method (modified)

from math import *

n = int(input("Enter number of rows: "))
p = int(input("Enter number of columns: "))

X = []

print("Enter matrix:")
for i in range(n):
    X.append(list(map(float, input().split())))

x_copy = X.copy()

V = [[0] * p for i in range(n)]
T = [[0] * p for i in range(p)]

for k in range(p):

    # Find norm
    s = 0
    for i in range(n):
        s = s + x_copy[i][k] ** 2

    d = sqrt(s)
    
    if d < e :
        print("Matrix is not full column rank")
        break
        
    # Find unit vector
    for i in range(n):
        V[i][k] = x_copy[i][k] / d

    # Find T
    for j in range(k, p):
        s = 0
        for i in range(n):
            s = s + V[i][k] * X[i][j]

        T[k][j] = s

    # Make remaining columns orthogonal
    for j in range(k + 1, p):
        for i in range(n):
            x_copy[i][j] = x_copy[i][j] - T[k][j] * V[i][k]

print("Matrix V:")
for i in range(n):
    for j in range(p):
        print(round(V[i][j], 4), end="\t")
    print()

print("Matrix T:")
for i in range(p):
    for j in range(p):
        print(round(T[i][j], 4), end="\t")
    print()
```

```python
# Householder Transformation

import numpy as np

n =  int(input("Enter the number of rows: "))
p =  int(input("Enter the number of columns: "))

x = np.zeros((n,p))

for i in range(n):
    for j in range(p):
        x[i,j] = float(input(f"enter x[{i+1},{j+1}]"))

print("original matrix x: ")
print(x)

for k in range(min(n,p)):
    s_sq = np.sum(x[k:, k] ** 2)

    if s_sq < e:
        continue

    s = np.sqrt(s_sq)

    u = x[k:, k].copy()

    if x[k,k] < 0:
        u[0] -= s

    else:
        u[0] += s

    beta = np.dot(u,u)

    for j in range(k+1, p):
        temp = np.dot(u, x[k:, j])
        x[k:, j ] -= (u* temp)/beta

    if x[k,k] < 0:
        x[k,k] = s

    else:
        x[k,k] = -s

    x[k+1: ,k] = 0

x[np.abs(x) < 1e-10 ] = 0

print("The transformed matrix is: ")
print(np.round(x,4))
```

```python
# Gauss Seidel for homogenous equations

import numpy as np

n = int(input("Enter the number of unknowns: "))

a = np.zeros((n,n))

print("enter the coefficient matrix a: ")
for i in range(n):
    for j in range(n):
        a[i,j] = float(input(f" a{i+1}{j+1}: "))

print("coefficient matrix a: ")
print(a)

b = np.zeros((n,1))

for i in range(n):
    b[i] = float(input(f"Enter b{i+1}: "))

print("right side vector b: ")
print(b)

diagonal_dominant = True
for i in range(n):
    diag_element = abs(a[i,i])
    off_diagonal = 0

    for j in range(n):
        if j != i:
            off_diagonal += abs(a[i,j])

    if diag_element <= off_diagonal:
        diagonal_dominant = False

if diagonal_dominant:
    print("Matrix A is diagonally dominant.")
    print("Gauss-Seidel method is expected to converge.")

else:
    print("Matrix A is not diagonally dominant.")
    print("convergence is not guaranteed.")


x = np.zeros(n)
eps = float(input("enter the accuracy level"))
N = int(input("Enter the maximum number of iteration: "))

for k in range(N):
    x_old = x.copy()

    for i in range(n):
        p = b[i,0]

        for j in range(n):
            if j != i : 
                p = p - a[i,j] * x[j]

        x[i] = p / a[i,i]

    error =  np.max(np.abs(x-x_old))

    print(f"Iteration {k+1}: {x}")

    if error < eps:
        print("Gauss-Seidel method converged.")
        break
       
else:
    print("Maximum iterations reached.")

print("\nSolution: ")

for i in range(n):
    print(f"x{i+1} = {x[i]}")
```

```python
# Gauss Elimination for Homogenous Equations

import numpy as np

n = int(input("Enter number of unknowns: "))

aug = np.zeros((n,n+1))

print("Enter the augmented matrix: ")
for i in range(n):
    for j in range(n+1):
        aug[i,j] = float(input(f"aug[{i+1},{j+1}] = "))

print("\n Augmented Matrix: ")
print(aug)

for k in range(n-1):
    if aug[k,k] == 0:
        for i in range(k+1, n):
            if aug[i, k] != 0:
                a[[k, i]] = a[[i, k]]
                break

    for i in range(k+1, n):
        m = aug[i,k]/aug[k,k]
        aug[i] = aug[i] - m * aug[k]

print("\nUpper triangular matrix: ")
print(aug)

x = np.zeros(n)

for i in range(n-1, -1, -1):
    s = 0

    for j in range(i+1, n):
        s += aug[i,j] * x[j]

    x[i] = (aug[i,n] - s) / aug[i,i]

print("\nSolution: ")
for i in range(n):
    print(f"x{i+1} = {x[i]}")
```

```python
# GAUSS JORDAN FOR MATRIX INVERSE

import numpy as np

n = int(input("Enter order of matrix: "))

# Enter matrix
A = np.zeros((n, n))

print("Enter matrix A:")
for i in range(n):
    for j in range(n):
        A[i, j] = float(input(f"A[{i+1},{j+1}] = "))

# Augment A with identity matrix
aug = np.hstack((A, np.eye(n)))

# Gauss-Jordan elimination
for i in range(n):

    # Check zero pivot
    if aug[i, i] == 0:
        for k in range(i + 1, n):
            if aug[k, i] != 0:
                aug[[i, k]] = aug[[k, i]]
                break
        else:
            print("Inverse does not exist.")
            exit()

    # Make pivot = 1
    aug[i] = aug[i] / aug[i, i]

    # Make other elements in pivot column = 0
    for j in range(n):
        if i != j:
            m = aug[j, i]
            aug[j] = aug[j] - m * aug[i]

# Extract inverse
inverse = aug[:, n:]

print("\nInverse of A:")
print(inverse)
```

```python
# SWEEP METHOD

n = int(input("Number of equations: "))

np.set_printoptions(precision=4, suppress=True)

# Input coefficient matrix A
A = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        A[i, j] = float(input(f"Enter A{i+1}{j+1}: "))

print("\nCoefficient matrix A =")
print(A)

# Input matrix b
b = np.zeros((n, 1))

for i in range(n):
    b[i, 0] = float(input(f"Enter b{i+1}: "))

print("\nMatrix b =")
print(b)

# Create augmented matrix [A | b]
aug_mat = np.hstack([A, b])

print("\nAugmented matrix [A | b] =")
print(aug_mat)

# Sweep method
for k in range(n):

    # Pivoting
    if np.abs(aug_mat[k, k]) == 0.0:
        for i in range(k + 1, n):
            if aug_mat[i, k] != 0:
                aug_mat[[k, i], :] = aug_mat[[i, k], :]
                print("Row", k+1, "and row", i+1, "interchanged.")
                break

    # Make pivot element 1
    aug_mat[k, :] = aug_mat[k, :] / aug_mat[k, k]

    # Make other elements in pivot column zero
    for i in range(n):
        if i == k or aug_mat[i, k] == 0.0:
            continue

        aug_mat[i, :] = aug_mat[i, :] - aug_mat[i, k] * aug_mat[k, :]

    # Remove small numerical errors
    aug_mat[np.abs(aug_mat) < 1e-10] = 0

    print("\nAfter sweep", k+1, ":")
    print(aug_mat)

# Solution
x = aug_mat[:, n:]
print("\nSolution:")
print(x)
```
