<h2>Conversion from other number system to decimal</h2>

```python
# Binary to Decimal

binary = input("enter a binary digit: ")
digit = 0

for i in binary:
    digit = digit*2 + int(i)
print(digit)
```

```python
# Octal to Decimal

octal = input("enter a binary digit: ")
d = 0

for i in octal:
    d = d*8 + int(i)
print(d)
```

```python
# Hexadecimal to Decimal

hexa = input("enter a hex digit: ").upper()
ref = "0123456789ABCDEF"
decimal = 0

for i in hexa:
    value = ref.index(i)
    decimal = decimal*16 + value

print(decimal)
```

```python
# fractional hexadecimal to decimal

number = input("enter a fractional hexadecimal of your choice: ").upper()
reference = "0123456789ABCDEF"
real = 0
fractional = 0

real_number, frac_number = number.split(".")

for i in real_number:
    value = reference.index(i)
    real = real*16 + value

power = 16

for k in frac_number:
    value = reference.index(k)
    fractional = fractional + (value/power)
    power *= 16

print(f"{real + fractional}")
    
```

```python
# fractional octal to decimal

number = input("enter a fractional octal of your choice: ")

real_number, frac_number = number.split(".")
real = 0
frac = 0
value = 0
for i in real_number:
    value = int(i)
    real = real*8 + value

power = 8

for k in frac_number:
    value = int(k)
    frac = frac + (value/power)
    power *= 8

print(f"{real + frac}")
```
