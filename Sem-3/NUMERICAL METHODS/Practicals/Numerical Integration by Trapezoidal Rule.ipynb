```python
# 1st Implementation

import math

def f(x):
    return math.sin(x)

def trapezoidal(f, a, b, n):
    h = (b - a) / n
    s = f(a) + f(b)

    for i in range(1, n):
        s += 2 * f(a + i * h)

    return h * s / 2


def simpson(f, a, b, n):
    if n % 2:
        raise ValueError("n must be even")

    h = (b - a) / n
    s = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h
        s += (4 if i % 2 else 2) * f(x)

    return h * s / 3


a, b, n = 0, math.pi, 10

print("Trapezoidal:", trapezoidal(f, a, b, n))
print("Simpson's:", simpson(f, a, b, n))
```

```python
# Simple Implementation

import math

a, b, n = 0, math.pi, 10
h = (b - a) / n

# Trapezoidal
s = math.sin(a) + math.sin(b)
for i in range(1, n):
    s += 2 * math.sin(a + i*h)
print("Trapezoidal =", h*s/2)

# Simpson's
s = math.sin(a) + math.sin(b)
for i in range(1, n):
    s += (4 if i % 2 else 2) * math.sin(a + i*h)
print("Simpson's =", h*s/3)
```
