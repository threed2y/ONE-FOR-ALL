```python
# Method 2 : Decimal to Binary Conversion

n = int(input("please enter your number here: "))

l = list()

if n == 0:
    print(n)

else:
     while  n != 0:
         remainder = n % 2
         l.append(remainder)
         n = n // 2

     l.reverse()
     for element in l:
         print(element, end="")
```
