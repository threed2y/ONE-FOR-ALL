<h2> Addition and Substraction of Binary numbers</h2>

```python
# Addition of two binary number

# Simplest Approach
binary1 = input("enter the first binary number: ")
binary2 = input("enter the first binary number: ")

int1,int2 = int(binary1,2), int(binary2,2)
solution = int1 + int2

solution = bin(solution)[2:]
print(f"Required solution of the given binary digits is: {solution}")
```

```python
# Approach 2

binary1, binary2 = input("Enter binary numbers separated by comma: ").split(",")

binary1 = binary1[::-1]
binary2 = binary2[::-1]

carry = 0
answer = ""

for i, j in zip(binary1, binary2):

    total = int(i) + int(j) + carry

    if total == 0:
        answer += "0"
        carry = 0

    elif total == 1:
        answer += "1"
        carry = 0

    elif total == 2:
        answer += "0"
        carry = 1

    elif total == 3:
        answer += "1"
        carry = 1

if carry:
    answer += "1"

answer = answer[::-1]

print("Answer:", answer)
```

```python
# Addition of two fractional binary numbers

a, b = input("Enter two binary fractions: ").split(",")

a = a.replace(".", "")
b = b.replace(".", "")

carry = 0
answer = ""

for i, j in zip(a[::-1], b[::-1]):
    total = int(i) + int(j) + carry

    answer += str(total % 2)
    carry = total // 2

if carry:
    answer += "1"

answer = answer[::-1]

print(answer)
```
