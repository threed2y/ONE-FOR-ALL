## Question 1

```python
import numpy as np

np.random.seed(42)

mu, sigma = 10, 2 
s = np.random.normal(mu, sigma, 50)
print(f"The generated random numbers are: \n {s}")

mean = np.mean(s)
print(f"The mean of the generated random numbers is: {mean:.4f}")

variance = np.var(s)
print(f"The variance of the generated random numbers is: {variance:.4f}")

sd = np.std(s)
print(f"The standard deviation of the generated random numbers is: {sd:.4f}")

maximum = np.max(s)
print(f"The maximum of generated random numbers is : {maximum:.4f}")

minimum = np.min(s)
print(f"The minimum of generated random numbers is : {minimum:.4f}")
```

## Question 2:

```python
array1 = np.array([[1,2], 
                   [3,4]])
print(f"Your first matrix is: \n {array1}")

array2 = np.array([[5,6], 
                   [7,8]])
print(f"Your second matrix is: \n {array2}")

print(f"The element wise addition of the arrays is: \n {array1 + array2}")
print(f"The element wise multiplication of the arrays is: \n {array1 * array2}")
print(f"The dot product of the arrays is: \n {np.dot(array1, array2)}")
print(f"The square of each element of first array is: \n {array1**2}")
print(f"The square of each element of second array is: \n {array2**2}")
```

## Question 3:

```python
matrix = []

print("Enter values for a 3x3 matrix:")

for i in range(3):
    row = list(map(int, input(f"Enter 3 numbers for row {i+1} separated by space: ").split()))
    matrix.append(row)

matrix = np.array(matrix)

print()

print("Matrix:")
print(matrix)
print()

transpose = np.transpose(matrix)
print(f"The transpose of the given matrix is: \n {transpose}")

print()

determinant = np.linalg.det(matrix)
print(f"The determinant of the given matrix is: {determinant}")

print()

inverse = np.linalg.inv(matrix)
print(f"The inverse of the given matrix is: \n {inverse}")
```

## Question 4:

```python
np.random.seed(69)

matrix2 = np.random.randint(0, 100, size = (4, 5))
print(f"The 4X5 matrix is: \n {matrix2}")

print()

second_row = matrix2[1:2]
print(f"The second row of the matrix is: \n {second_row}")

print()

third_column = matrix2[:,2:3]
print(f"The third column of the matrix is: \n {third_column}")

print()

submatrix = matrix2[1:3, 1:3]
print(f"The submatrix from the center is: \n {submatrix}")
```
