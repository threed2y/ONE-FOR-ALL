## Question 5.

```python
num = int(input("How many numbers do you want to enter?"))

my_list = []

even = 0
odd = 0

for i in range(num):
    number_input = int(input("Please enter the number: "))
    my_list.append(number_input)
    if number_input % 2 == 0:
        even += 1
    else:
        odd += 1

maximum = my_list[0]
minimum = my_list[0]   

for i in my_list:
    if i > maximum:
        maximum = i
    if i < minimum:
        minimum = i
        
print(f"The list of numbers you have input is {my_list}")
print(f"The sum of the inputted numbers is {sum(my_list)}.")

odd_list = my_list[1::2]
even_list = my_list[::2]

print(f"The numbers inputted at odd positions are {odd_list}")
print(f"The numbers inputted at even positions are {even_list}")

print(f"The number of odd numbers inputted are {odd}")
print(f"The number of even numbers inputted are {even}")

print(f"The maximum of inputted numbers is {maximum}")
print(f"The minimum of inputted numbers is {minimum}")

new_list = list(set(my_list))
print(f"The list without any duplicate value is {new_list}")
```

## Question 6.

```python
l1 = [53, 21, 62, 92, 12, 60]
l2 = []
item_check = int(input("How many items do you want to check?"))

for i in range(item_check):
    number = int(input("What is the number you want to check?"))
    l2.append(number)
    if number in l1:
        print(f"{number} is present is the list.")
    else:
        print(f"{number} is not present is the list.")

perm = input("Do you want to add the missing numbers in the list? Y/N")

if perm.lower() == "y":
    for num in l2:
        if num not in l1:
            l1.append(num)
print(f"Updated list: {l1}")
```

## Question 7.

```python
words = ["madam", "apple", "level", "banana", "civic", "grape"]

for word in words:
    if word == word[::-1]:
        print(f"{word} is a palindrome")
    else:
        print(f"{word} is not a palindrome")
```
