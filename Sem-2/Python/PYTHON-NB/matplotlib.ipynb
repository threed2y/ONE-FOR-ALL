## 1. Examine the distribution of generated marks

```python
import random
import numpy as np
import matplotlib.pyplot as plt

random.seed(27)

marks_stats = np.random.normal(loc = 25, scale = 5, size = 50)
marks_maths = np.random.normal(loc = 18, scale = 15, size = 50)

plt.hist(marks_stats, bins=30, edgecolor='black', alpha=0.7)
plt.title("Marks of statistics")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

plt.hist(marks_maths, bins=30, color = "yellow", edgecolor='black', alpha=0.7)
plt.title("Marks of Mathematics")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import random
import numpy as np
import matplotlib.pyplot as plt

random.seed(27)

marks_stats = np.clip(np.random.normal(loc=25, scale=5, size=50), 0, None)
marks_maths = np.clip(np.random.normal(loc=18, scale=15, size=50), 0, None)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].hist(marks_stats, bins=30, edgecolor='black', alpha=0.7)
ax[0].set_title("Marks of Statistics")
ax[0].set_xlabel("Marks")
ax[0].set_ylabel("Frequency")

ax[1].hist(marks_maths, bins=30, color="yellow", edgecolor='black', alpha=0.7)
ax[1].set_title("Marks of Mathematics")
ax[1].set_xlabel("Marks")
ax[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

## 2. Compare the sales and identify the pattern

```python
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul","Aug", "Sep", "Oct", "Nov", "Dec"])
product_A = np.array([25, 26.7, 27.2, 28.9, 22.3, 25.2, 27.5, 30.1, 32.9, 33.4, 34, 35.02])
product_B = np.array([40.3, 37, 36, 32, 28.9, 26.3, 22.4, 21.2, 29.8, 31.5, 32.1, 36.7])

plt.plot(months, product_A, color = "Black", label = "Product A")
plt.plot(months, product_B, color = "Red", label = "Product B")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show();
```

```python
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul","Aug", "Sep", "Oct", "Nov", "Dec"])
product_A = np.array([25, 26.7, 27.2, 28.9, 22.3, 25.2, 27.5, 30.1, 32.9, 33.4, 34, 35.02])
product_B = np.array([40.3, 37, 36, 32, 28.9, 26.3, 22.4, 21.2, 29.8, 31.5, 32.1, 36.7])

plt.figure(figsize=(16, 8))

plt.plot(months, product_A, color="black", marker="o", linewidth=2, label="Product A")
plt.plot(months, product_B, color="red", marker="s", linestyle="--", linewidth=2, label="Product B")

plt.title("Monthly Sales Comparison", fontsize=14, fontweight="bold")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

## 3. Simulation of Data

```python
random.seed(96)

salary = np.clip(np.random.normal(loc = 50000, scale = 20000, size = 500), 10000, None)

fig, ax = plt.subplots(1, 2, figsize=(16, 8))

ax[0].hist(salary, bins=100, edgecolor='black', alpha=0.7)
ax[0].set_title("Income Distribution")
ax[0].set_xlabel("Income")
ax[0].set_ylabel("Number of people")
ax[0].set_xticks(np.arange(0, 130000, 20000))

ax[1].boxplot(salary, orientation = "horizontal", meanline= True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import random

random.seed(96)

salary = np.clip(np.random.normal(loc=50000, scale=20000, size=500), 10000, None)

mean_val = np.mean(salary)
median_val = np.median(salary)

fig, ax = plt.subplots(1, 2, figsize=(12, 8), sharex=True)

# Histogram
ax[0].hist(salary, bins=40, color="#457b9d", edgecolor='black', alpha=0.8)
ax[0].axvline(mean_val, color='red', linestyle='--', linewidth=1.5, label=f"Mean: {int(mean_val)}")
ax[0].axvline(median_val, color='green', linestyle='-', linewidth=1.5, label=f"Median: {int(median_val)}")

ax[0].set_title("Income Distribution", fontsize=14, fontweight="bold")
ax[0].set_ylabel("Number of People")
ax[0].legend()
ax[0].grid(axis='y', linestyle='--', alpha=0.4)

# Boxplot
ax[1].boxplot(
    salary,
    vert=False,
    patch_artist=True,
    boxprops=dict(facecolor="#a8dadc"),
    medianprops=dict(color="green", linewidth=2),
    meanprops=dict(color="red", linewidth=2),
    showmeans=True
)

ax[1].set_xlabel("Income")
ax[1].grid(axis='x', linestyle='--', alpha=0.4)

# Format x-axis
for a in ax:
    a.xaxis.set_major_formatter(lambda x, _: f'{int(x):,}')

plt.xlim(0, salary.max())
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import random

random.seed(96)

# Generate salary data
salary = np.clip(np.random.normal(loc=50000, scale=20000, size=500), 10000, None)

plt.figure(figsize=(10,5))

# Compute histogram
counts, bins, patches = plt.hist(salary, bins=40, edgecolor='black', alpha=0.7)

# Compute bin centers for line
bin_centers = 0.5 * (bins[1:] + bins[:-1])

# Overlay line
plt.plot(bin_centers, counts, color="red", linewidth=2, label="Histogram Line")

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of People")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
