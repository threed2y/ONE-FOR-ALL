```python
# Q 3 -- STUDY OF INCOME DISTRIBUTION

income = np.random.normal(50000, 20000, size = 500)
income_exp = np.random.exponential(50000, size = 500)
x_vals = np.linspace(min(income), max(income), 100)
x_v = np.linspace(min(income_exp), max(income_exp), 100)


plt.hist(income, bins=10, density=True, alpha=0.6, color='cyan', edgecolor = "black")
# Density curve
density = gaussian_kde(income)
plt.plot(x_vals, density(x_vals), color='red', linewidth=2)
plt.xlabel("Sales")
plt.ylabel("Density")
plt.title("Histogram with Normal Density Curve")
plt.show()


plt.hist(income_exp, bins=10, density=True, alpha=0.6, color='pink', edgecolor = "black")
# Density curve
density = gaussian_kde(income_exp)
plt.plot(x_v, density(x_v), color='blue', linewidth=2)
plt.xlabel("Sales")
plt.ylabel("Density")
plt.title("Histogram with Exponential Density Curve")
plt.show()
```
