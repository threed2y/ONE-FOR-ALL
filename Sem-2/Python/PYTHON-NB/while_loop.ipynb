```python
## Question 1. Write countdown
```

```python
from datetime import datetime
num = int(input("Please enter a number to begin countdown: "))

start = datetime.now()

if  num > 0:
    while num >= 0:
        if num > 100:
            print(num)
            num -= 10
        elif 0 <= num <= 100:
            print(num)
            num -= 1
else:
    print("Please enter a positive number to begin.")


stop = datetime.now()
print(f"The time taken to perform the countdown is {(datetime.now() - start).total_seconds()} seconds.")
```

```python
## Question 2. Character frequency distribution
```

```python
input_string = input("Please enter a string: ")
char_freq = {}

for char in input_string.lower():
    if char in char_freq:
        char_freq[char] += 1
    else:
        char_freq[char] = 1

print(f"The frequency distribution of characters in the string is {char_freq}")
```

```python
input_str = input("Please enter a string: ")

i = 0
n = len(input_str)
visited = ""

while i < n:
    char = input_str[i]
    if char not in visited:
        count = 0
        visited += input_str[i]
        j = 0

        while j < n:
            if input_str[j] == char:
                count += 1
            j += 1

        print(input_str[i], ":", count)

    i += 1
```

```python
## Question 3. String Cleaning
```

```python
import string
s = "A1$b2_C3d4*E5f6#G7h8!9@0"
list_1 = []
list_2 = []

i = 0
n = len(s)

while i < n:
    if s[i] in string.ascii_lowercase or s[i] in string.ascii_uppercase:
        list_1.append(s[i])
    else:
        list_2.append(s[i])
    i += 1
print(f"The list containing alphabets is {list_1}")
print(f"The list containing other  is {list_2}")
```

```python
##  Question 4. Random number guessing
```

```python
import random

comp_rand = random.randint(1,25)
i = 1
guessed = False

while i <= 3:
    user_input = int(input(f"This is your attempt {i}, what is your guess?"))
    if user_input == comp_rand:
        print("Congratulations, you have guessed the correct number!")
        guessed = True
        break
    else:
        if i < 3:
            print("Please try again.")
    i += 1

if not guessed:
    print("You have failed to guess the number correctly.")
    print(f"The number chosen by computer was {comp_rand}")
```

```python
## Question 5. Subsequence check
```

```python
s2 = input("Please enter your main string: ").lower()
s1 = input("Please enter the subsequence you want to check: ").lower()

i, j = 0, 0

while i < len(s2) and j < len(s1):
    if s2[i] == s1[j]:
        j += 1
    i += 1

sub = (j == len(s1))
print(sub)
```
